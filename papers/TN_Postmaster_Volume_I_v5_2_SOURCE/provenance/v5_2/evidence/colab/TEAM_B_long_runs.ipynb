## A. High-precision recheck

f(s) = Σ aₙ n⁻ˢ, where aₙ is periodic mod 5 with values (1, κ, −κ, −1, 0) and κ = (√(10−2√5) − 2)/(√5 − 1).
The completed function is F(s) = (5/π)^{s/2} Γ((s+1)/2) f(s), and it should satisfy F(s) = F(1−s).

```
K = (mp.sqrt(10 - 2*mp.sqrt(5)) - 2) / (mp.sqrt(5) - 1)
COEF = {1: 1, 2: K, 3: -K, 4: -1}
def f(s): return mp.power(5, -s) * sum(c * mp.zeta(s, mp.mpf(r)/5) for r, c in COEF.items())
def F(s): return mp.power(5/mp.pi, s/2) * mp.gamma((s+1)/2) * f(s)
for s in [mp.mpc(0.3, 40), mp.mpc(1.2, 97), mp.mpc(-0.2, 150)]:
    print('functional equation |F(s)/F(1-s) - 1| =', mp.nstr(abs(F(s)/F(1-s) - 1), 5))
off = [mp.findroot(f, mp.mpc(s, t)) for s, t in ZEROS['offline']]
for z in off: print('off-line zero', mp.nstr(z, 25), ' |f| =', mp.nstr(abs(f(z)), 3))
```

```
# Polish the on-line zeros near x* (those with |log(q/x*)| < 0.3) and recompute the sign of W_k.
def Zt(t):
    s = mp.mpc(0.5, t)
    th = mp.im(s/2*mp.log(5/mp.pi) + mp.loggamma((s+1)/2))
    return mp.re(mp.expj(th) * f(s))
xstar = mp.mpf(FIRST[0]['x']); kstar = FIRST[0]['kFirst']
near = [g for g in ZEROS['online'] if abs(math.log((0.25+g*g)/float(xstar))) < 0.3]
on_hp = [mp.findroot(Zt, mp.mpf(g)) for g in near]
print(len(on_hp), 'nearby on-line zeros polished; max shift', mp.nstr(max(abs(a-b) for a, b in zip(on_hp, near)), 3))
orbits = [(mp.mpf(1)/4 + g*g, 1) for g in on_hp] + [(z*(1-z), 2) for z in off]
def G(x, k):  # sign of W_k(x); scaled by the largest |4x u|^k
    terms = [(4*x*q/(x+q)**2, w) for q, w in orbits]
    m = max(abs(u) for u, w in terms)
    return sum(w * mp.re((u/m)**k) for u, w in terms)
def golden_min(fn, a, b, it=120):
    g = (mp.sqrt(5)-1)/2; c = b-g*(b-a); d = a+g*(b-a); fc, fd = fn(c), fn(d)
    for _ in range(it):
        if fc < fd: b, d, fd = d, c, fc; c = b-g*(b-a); fc = fn(c)
        else: a, c, fc = c, d, fd; d = a+g*(b-a); fd = fn(d)
    return (c, fc) if fc < fd else (d, fd)
for k in [kstar-1, kstar]:
    xm, gm = golden_min(lambda x: G(x, k), xstar*mp.mpf('0.998'), xstar*mp.mpf('1.002'))
    print(f'k={k}: min over x in x*(1 +- 0.002) of scaled W_k sign function = {mp.nstr(gm, 8)} at x = {mp.nstr(xm, 15)}')
```

## B. Higher zero list and the k*·θ² law (long run)

The script is the same as the local `zeros.py` (argument principle in 5-unit windows, on-line zeros from sign changes of Z, off-line zeros from |f| minima plus Newton). It runs in mpmath at 20 digits.
For each off-line zero it records θ = |arg q| and the first failing rung k\* near its |q|. The prediction to test is **k\*·θ² ≈ c**, with c between about 0.2 and 1.5 depending on how the nearby on-line zeros are spaced.

```
mp.mp.dps = 20
def count(t0, t1, SL=-0.45, SR=1.45, n=40):
    pts = [mp.mpc(SL+(SR-SL)*i/n, t0) for i in range(n+1)] + [mp.mpc(SR, t0+(t1-t0)*i/n) for i in range(1, n+1)] \
        + [mp.mpc(SR-(SR-SL)*i/n, t1) for i in range(1, n+1)] + [mp.mpc(SL, t1-(t1-t0)*i/n) for i in range(1, n+1)]
    tot = 0; vals = [f(p) for p in pts]
    for a, b, fa, fb in zip(pts, pts[1:], vals, vals[1:]):
        stack = [(a, b, fa, fb)]
        while stack:
            a_, b_, fa_, fb_ = stack.pop(); d = mp.arg(fb_/fa_)
            if abs(d) > 0.4: m = (a_+b_)/2; fm = f(m); stack += [(m, b_, fm, fb_), (a_, m, fa_, fm)]
            else: tot += d
    return tot/(2*mp.pi)
T0, T1 = 260, 600    # extend beyond the local list; raise T1 for an overnight run
t = T0; new_online = []; new_off = []
while t < T1:
    tot = int(mp.nint(count(t, t+5)))
    ts = [t + 0.02*i for i in range(251)]; zs = [Zt(u) for u in ts]
    on = [mp.findroot(Zt, (a, b), solver='anderson') for a, b, za, zb in zip(ts, ts[1:], zs, zs[1:]) if za*zb < 0]
    new_online += on
    if tot > len(on):
        print(f'window {t}-{t+5}: {tot} zeros, {len(on)} on-line -> {tot-len(on)} off-line to locate')
        new_off.append((t, tot-len(on)))
    t += 5
print('extended list:', len(new_online), 'on-line zeros; windows with off-line zeros:', new_off)
```

## C. CUE baseline for the Project 51 next-gap classifier

Project 51 v8 reports log-score gains of 0.038–0.055 bits for predicting "next unfolded gap > 1" from the previous gap (bins at 0.5, 1, 1.5), falling with height.
Random-matrix theory predicts a limiting value for this exact task. This section estimates it from Haar-unitary (CUE) spectra.

```
import numpy as np
rng = np.random.default_rng(51)
def cue_gaps(N):
    Zm = (rng.standard_normal((N, N)) + 1j*rng.standard_normal((N, N)))/np.sqrt(2)
    Q, R = np.linalg.qr(Zm); Q = Q * (np.diag(R)/abs(np.diag(R)))
    th = np.sort(np.angle(np.linalg.eigvals(Q)))
    g = np.diff(np.concatenate([th, [th[0]+2*np.pi]]))*N/(2*np.pi)
    return g
prev, nxt = [], []
for _ in range(4000):
    g = cue_gaps(100); prev.append(g[:-1]); nxt.append(g[1:])
prev = np.concatenate(prev); nxt = np.concatenate(nxt)
bins = np.searchsorted([.5, 1, 1.5], prev, side='right'); y = (nxt > 1).astype(int)
p = np.array([y[bins == b].mean() for b in range(4)]); q0 = y.mean()
q = p[bins]; gain = np.mean(np.where(y, np.log2(q/q0), np.log2((1-q)/(1-q0))))
print('CUE P(next>1 | prev bin) =', np.round(p, 4), ' base', round(q0, 4))
print('CUE limiting gain (bits) =', round(gain, 5), ' vs Project 51 observed 0.038-0.055')
```

## D. Principal route: the smallest failing Löwner size N\* for the twin (high precision)

By the Dobsch–Donoghue theorem, every N-node Löwner matrix of h(x) = x·S(x) is PSD exactly when, at every centre x₀, the N×N matrix [h^(i+j+1)(x₀)/(i+j+1)!] is PSD.

Here the Taylor coefficients come straight from the twin's source formula, S(x) = Λ′/Λ(s_x)/R, sampled on a circle in the x-plane. So every zero, including those above any tabulated height, is included exactly. Inertia is read off by symmetric elimination at 250 digits.

In double precision, the local run saw no failure through N = 64. The smallest eigenvalue had fallen to 10⁻¹⁵ by N ≈ 65, and the drop steepened after N ≈ 43, the number of on-line zeros below the off-line one. This section finds the actual first failing N, or pushes the lower bound up.

```
mp.mp.dps = 250
K = (mp.sqrt(10 - 2*mp.sqrt(5)) - 2) / (mp.sqrt(5) - 1)
COEF = {1: 1, 2: K, 3: -K, 4: -1}
def fD(s):  return mp.power(5, -s) * sum(c * mp.zeta(s, mp.mpf(r)/5) for r, c in COEF.items())
def fDp(s): return -mp.log(5)*fD(s) + mp.power(5, -s) * sum(c * mp.zeta(s, mp.mpf(r)/5, 1) for r, c in COEF.items())
def h_twin(x):
    R = mp.sqrt(1 + 4*x); s = (1 + R)/2
    return x * (mp.log(5/mp.pi)/2 + mp.digamma((s+1)/2)/2 + fDp(s)/fD(s)) / R
def taylor_coeffs(x0, nmax, M=2048, frac=mp.mpf('0.75')):
    r = frac * x0
    vals = [h_twin(x0 + r*mp.expj(2*mp.pi*k/M)) for k in range(M)]
    out = []
    for n in range(nmax + 1):
        c = sum(vals[k]*mp.expj(-2*mp.pi*k*n/M) for k in range(M)) / M
        out.append(mp.re(c) * (x0/r)**n)          # coefficient of ((x-x0)/x0)^n
    return out
def negative_pivots(A):
    A = [row[:] for row in A]; n = len(A); neg = 0; smallest = None
    for k in range(n):
        p = A[k][k]
        if smallest is None or abs(p) < abs(smallest): smallest = p
        if p < 0: neg += 1
        for i in range(k+1, n):
            m = A[i][k] / p
            for j in range(k+1, n): A[i][j] -= m * A[k][j]
    return neg, smallest
q0 = off[0]*(1 - off[0]); r0 = abs(q0)
for frac_x in ['0.97', '1.00', '1.03']:
    x0 = r0 * mp.mpf(frac_x); t0 = time.time()
    c = taylor_coeffs(x0, 2*160)
    print(f'x0 = {mp.nstr(x0, 8)}: coefficients in {time.time()-t0:.0f}s')
    for N in range(40, 161, 5):
        A = [[c[i+j+1] for j in range(N)] for i in range(N)]
        neg, pmin = negative_pivots(A)
        print(f'   N={N:3d}: negative pivots {neg}   smallest |pivot| {mp.nstr(abs(pmin), 5)}')
        if neg: break
```
